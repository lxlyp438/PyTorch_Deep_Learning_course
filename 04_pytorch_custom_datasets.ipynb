```python
import torch
from torch import nn

device = "cuda" if torch.cuda.is_available() else "cpu"

# load some data Food-101 data set - start with a subset
# load data from github
import os
import requests
import zipfile
from pathlib import Path

# setup path to a data folder
data_path = Path("data/")
image_path = data_path / "food101_subset"

# if the image folder not exist, download it and prepare it ...
if image_path.is_dir():
    print(f"{image_path} directory exists")
else:
    print(f"Did not find {image_path} directory, creating one...")
    image_path.mkdir(parents=True, exist_ok=True)

# download food101_subset data
with open(data_path / "food101_subset.zip", "wb") as f:
    request = requests.get("https://github.com/lxlyp438/PyTorch_Deep_Learning_course/raw/refs/heads/main/data/food101_subset.zip")
    print("Downloading food101_subset.zip ...")
    f.write(request.content)

# unzip food101_subset data
with zipfile.ZipFile(data_path / "food101_subset.zip", "r") as zip_ref:
    print("Unzipping food101_subset.zip ...")
    zip_ref.extractall(image_path)
```
